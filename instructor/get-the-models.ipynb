# Get the Models

Run this once before class. It downloads the models into a shared folder so every student uses the same copy.

Students do not run this notebook.

## The lineup

One family, four sizes. Same company, same chat template, so the only thing that changes is size.

| Model | Parameters | Download |
|---|---|---|
| Qwen3.5-0.8B | 0.8 billion | 0.5 GB |
| Qwen3.5-2B | 2 billion | 1.3 GB |
| Qwen3.5-4B | 4 billion | 2.7 GB |
| Qwen3.5-9B | 9 billion | 5.7 GB |

These are 4-bit versions (`Q4_K_M`) from [unsloth](https://huggingface.co/unsloth) on Hugging Face. Apache 2.0 license.

```python
MODELS = {
    "Qwen3.5-0.8B": ("unsloth/Qwen3.5-0.8B-GGUF", "Qwen3.5-0.8B-Q4_K_M.gguf"),
    "Qwen3.5-2B":   ("unsloth/Qwen3.5-2B-GGUF",   "Qwen3.5-2B-Q4_K_M.gguf"),
    "Qwen3.5-4B":   ("unsloth/Qwen3.5-4B-GGUF",   "Qwen3.5-4B-Q4_K_M.gguf"),
    "Qwen3.5-9B":   ("unsloth/Qwen3.5-9B-GGUF",   "Qwen3.5-9B-Q4_K_M.gguf"),
}
```

## Where they go

On a hub, instructors can write to `shared-readwrite`. Students see the same folder as `shared`, read only.

On your laptop, they go in a `models` folder next to this notebook.

```python
from pathlib import Path

hub = Path("/home/jovyan/shared-readwrite")
folder = hub / "open-models" if hub.exists() else Path("models")
folder.mkdir(parents=True, exist_ok=True)
folder
```

## Download

This takes a few minutes. Files that are already there get skipped.

```python
from huggingface_hub import hf_hub_download

for name, (repo, file) in MODELS.items():
    path = Path(hf_hub_download(repo_id=repo, filename=file, local_dir=folder))
    print(f"{name:14} {path.stat().st_size / 1e9:.1f} GB")
```

## Will it fit?

Our GPU hubs use an NVIDIA T4 with 16 GB, shared by two students. The GPU memory is not split between them, so both models have to fit at once.

The 9B model is 5.7 GB. Two of them is about 11 GB, which leaves room for the conversation. That is why we stop at 9B.
